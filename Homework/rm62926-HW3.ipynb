```python
import numpy as np
import math
import matplotlib.pyplot as plt

def taylorSeriesSin(x, terms):
    
    sin_value = 0.0
    
    for i in range(0, terms):
        sin_value += (((-1)**i) * ((x**(2*i + 1)) / (math.factorial(2*i + 1))))
        
    return sin_value
```

```python
x = np.arange(0, 10, 0.1)

y1 = np.arange(0, 10, 0.1)
y1_sin = np.arange(0, 10, 0.1)
y1_error = np.arange(0, 10, 0.1)

y2 = np.arange(0, 10, 0.1)
y2_sin = np.arange(0, 10, 0.1)
y2_error = np.arange(0, 10, 0.1)

y3 = np.arange(0, 10, 0.1)
y3_sin = np.arange(0, 10, 0.1)
y3_error = np.arange(0, 10, 0.1)

y4 = np.arange(0, 10, 0.1)
y4_sin = np.arange(0, 10, 0.1)
y4_error = np.arange(0, 10, 0.1)

# y_test = np.arange(0, 10, 0.1)
# y_test_sin = np.arange(0, 10, 0.1)
# y_test_error = np.arange(0, 10, 0.1)

for i in range(0, len(x)):
    y1[i] = taylorSeriesSin(x[i], 1)
    y1_sin[i] = np.sin(x[i])
    
for i in range(0, len(x)):
    y2[i] = taylorSeriesSin(x[i], 2)
    y2_sin[i] = np.sin(x[i])

for i in range(0, len(x)):
    y3[i] = taylorSeriesSin(x[i], 3)
    y3_sin[i] = np.sin(x[i])

for i in range(0, len(x)):
    y4[i] = taylorSeriesSin(x[i], 4)
    y4_sin[i] = np.sin(x[i])
    
for i in range(0, len(x)):
    y_test[i] = taylorSeriesSin(x[i], 20)
    y_test_sin[i] = np.sin(x[i])
    
y1_error = y1 - y1_sin
y2_error = y2 - y2_sin
y3_error = y3 - y3_sin
y4_error = y4 - y4_sin
y_test_error = y_test - y_test_sin
```

```python
plt.plot(x, y1_error, label='1 term', marker='o')
plt.plot(x, y2_error, label='2 terms', marker='s')
plt.plot(x, y3_error, label='3 terms', marker='^')
plt.plot(x, y4_error, label='4 terms',marker='d')
plt.plot(x, y_test_error, label='20000 terms',marker='d')

plt.xlabel('x', fontsize=12)
plt.ylabel('Truncation Error (Taylor - Actual)')
plt.title('Truncation Error vs x for Different Numbers of Taylor Series Terms')
plt.legend()
plt.show()
```

Discuss how the truncation error behaves as you include more terms. Does the error decrease uniformly for all values of x? Does the truncation error behave differently for small vs. large values of x (try feeding larger values of x)? (a few sentences is fine)

The truncation error decreases with more terms, but the improvement is not uniform across all x values. The error is much smaller near x = 0 and grows rapidly for larger x values, the approximation becomes very poor for x > 5 because the polynomial terms grow unbounded while sin(x) oscillates between -1 and 1.
