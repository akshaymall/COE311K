```python
import random

# @param numSamples: number of samples to take to get an approximation
# @return: returns the pi approximation after numSamples iterations
def monteCarloApproximation(numSamples):
    counter = 0
    for i in range(numSamples):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        if (x*x + y*y < 1):
            counter += 1
            
    return 4.0 * counter / numSamples
```

```python
import numpy as np
import matplotlib.pyplot as plt

samples = np.arange(10, 1001, 20) #Creates an array of numbers between 10 and 1000 with a step size of 20
errorArr = np.zeros(len(samples)) #Creates an array of zeros of the same length as samples
for i in range(0, len(samples)): #Creates approximation of the samples and then computes the error and adds it to errorArr to plot
    approximation = monteCarloApproximation(samples[i]) 
    errorArr[i] = np.abs(approximation - np.pi)
```

```python
plt.plot(samples, errorArr)
plt.xlabel('Number of Samples')
plt.ylabel('Error between Approx. Pi and Actual Pi')
plt.title('Error between Approx. Pi & Actual Pi vs Number of Samples')
plt.show()
```
