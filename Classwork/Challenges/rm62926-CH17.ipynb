```python
import numpy as np
import matplotlib.pyplot as plt
# Define the parameters
k = 1.0 # Constant in the equation (e.g., springconstant)
h = 0.01 # Step size
x_start = 0 # Start of the interval
x_end = 10 # End of the interval
y_0 = 1.0 # Initial condition for y
v_0 = 0.0 # Initial condition for dy/dx

# Create arrays to store x, y, and v values
x_values = np.arange(x_start, x_end + h, h)
y_values = np.zeros(len(x_values))
v_values = np.zeros(len(x_values))

# Set initial conditions
y_values[0] = y_0
v_values[0] = v_0
```

```python
import numpy as np
import matplotlib.pyplot as plt

k = 1.0
h_values = [0.01, 0.02, 0.03, 0.05, 0.1, 0.2, 0.3, 0.5]

x_start = 0
x_end = 10
y_0 = 1.0
v_0 = 0.0

plt.figure(figsize=(12, 6))

for h in h_values:
    x_values = np.arange(x_start, x_end + h, h)
    y_values = np.zeros(len(x_values))
    v_values = np.zeros(len(x_values))

    y_values[0] = y_0
    v_values[0] = v_0

    for i in range(1, len(x_values)):
        y_values[i] = y_values[i-1] + h * v_values[i-1]
        v_values[i] = v_values[i-1] - h * k * y_values[i-1]

    plt.plot(x_values, y_values, label=f'h={h}')

plt.xlabel('x')
plt.ylabel('y(x)')
plt.title('Euler Method Stability for Harmonic Oscillator')
plt.legend()
plt.grid(True)
plt.show()
```

Euler’s method works reasonably well for very small step sizes, but for this oscillatory system it becomes visibly unstable around h = 0.2 – 0.3 (for k=1), where the amplitude starts growing instead of oscillating. It is becoming unstable as the solution gains energy and causes the oscillations to grow outward. When you increase k, the system oscillates faster, so instability appears at even smaller h; decreasing k delays it. For k=10 and h=0.01, the solution looks stable at first but still slowly grows. Overall, Euler’s Method cannot handle this system with larger h because it fails to preserve the bounded, oscillatory behavior. Moreover, larger step sizes make the instability happen faster.
