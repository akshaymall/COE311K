```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D
```

```python
x = np.linspace(0, 2 * np.pi, 50)
fig = plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1) # (row, column, active area)
x_tan = np.linspace(-1 * np.pi / 2, np.pi / 2, 50)
plt.plot(x_tan, np.arctan(x_tan), 'y')

plt.subplot(2, 2, 2)
plt.plot(x, np.cos(x), 'g') # will go into active area #2

# plt.subplot(2, 2, 3)
ax = fig.add_subplot(2, 2, 3, projection = '3d')
def f1(x,y):
    return (x*x+y*y)

x_mesh = np.linspace(-10, 10, 100)
y_mesh = np.linspace(-10, 10, 100)
X, Y = np.meshgrid(x_mesh, y_mesh)
Z = f1(X, Y)
ax.plot_surface(X, Y, Z, cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('x² + y²')

plt.subplot(2, 2, 4)
x_scatter = np.random.rand(1000)
y_scatter = np.random.rand(1000)
size = np.random.rand(1000) * 50
color = np.random.rand(1000)
plt.scatter(x_scatter, y_scatter, size, color)
plt.colorbar()
plt.show()
```
