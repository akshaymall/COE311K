```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import CubicSpline

# Define known data points
x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([0, 1, 0, 1, 0, 1])

# Create a cubic spline interpolation of the data
cs = CubicSpline(x, y, bc_type='natural')

# Generate new x values for a smooth plot
x_interp = np.linspace(0, 5, 100)
y_interp = cs(x_interp)

degree = 3
coeffs = np.polyfit(x, y, degree)
poly = np.poly1d(coeffs)
y_poly = poly(x_interp)
```

```python
fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(x_interp, y_interp, color='blue', label='Cubic Spline')
ax.plot(x_interp, y_poly,   color='red', label=f'Polynomial Fit (degree {degree})')
ax.scatter(x, y, color='black', label='Data points')

ax.set_title('Cubic Spline vs. Polynomial Fit')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()

plt.show()
```
