```python
import numpy
```

```python
def binaryConvertor(num):
    if num == 0:
        return "0"
    
    finalProduct = ""
    quotient = num
    remainder = []
    while quotient > 0:
        remainder.append(str(quotient % 2))
        quotient = quotient // 2
        
    for i in range(len(remainder) - 1, -1, -1):
        finalProduct += remainder[i]
    
    return finalProduct
```

```python
print(binaryConvertor(2025))
```
