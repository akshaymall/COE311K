```python
# Challenge A: Linear Least Squares vs Polynomial Fit

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

n = 20
X1 = np.sort(np.random.uniform(0, 10, n))
noise = np.random.normal(0, 1.0, n)
Y1 = 2.5 * X1 + 1.0 + noise

# Linear Least Squares
A1 = np.vstack([X1, np.ones(len(X1))]).T
m1, c1 = np.linalg.lstsq(A1, Y1, rcond=None)[0]

Y1_linear_pred = m1 * X1 + c1

# Linear Model
mse_linear = np.mean((Y1 - Y1_linear_pred)**2)
ss_total = np.sum((Y1 - np.mean(Y1))**2)
ss_res_linear = np.sum((Y1 - Y1_linear_pred)**2)
r2_linear = 1 - (ss_res_linear / ss_total)

print("Linear Fit")
print(f"Slope: {m1:.4f}, Intercept: {c1:.4f}")
print(f"MSE: {mse_linear:.4f}")
print(f"R^2: {r2_linear:.4f}")

# Polynomial Fit
degree = 3
coefficients = np.polyfit(X1, Y1, degree)
polynomial = np.poly1d(coefficients)

Y1_poly_pred = polynomial(X1)

# Polynomial Model
mse_poly = np.mean((Y1 - Y1_poly_pred)**2)
ss_res_poly = np.sum((Y1 - Y1_poly_pred)**2)
r2_poly = 1 - (ss_res_poly / ss_total)

print("\nPolynomial Fit (Degree 3)")
print("Coefficients:", coefficients)
print(f"MSE: {mse_poly:.4f}")
print(f"R^2: {r2_poly:.4f}")

# ------------------------
# Plot Comparison
# ------------------------
x_smooth = np.linspace(0, 10, 200)

plt.scatter(X1, Y1, label="Data")
plt.plot(x_smooth, m1 * x_smooth + c1, label="Linear Fit")
plt.plot(x_smooth, polynomial(x_smooth), label="Polynomial Fit (deg=3)")
plt.xlabel("X1")
plt.ylabel("Y1")
plt.title("Linear vs Polynomial Fit")
plt.legend()
plt.grid(True)
plt.show()
```
