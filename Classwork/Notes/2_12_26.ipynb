```python
A = np.array([[1, 2, 3], #2x3 matrix
              [4, 5, 6]])
print("Matrix A:\n", A)
```

```python
B = np.array([[1, 2, 3], #3x3 matrix
              [4, 5, 6],
              [7, 8, 9]])
print("Matrix B:\n", B)
```

```python
matrix_mul = np.dot(A, B) # A@B: dot product of the rows of the first matrix and the columns of the second matrix
print("A * B:\n", matrix_mul) 
#[[(1x1 + 2x4 + 3x7), (1x2 + 2x5 + 3x8), (...)]
#[(4x1 + 5x4 + 6x7), (...), (...)]]
```

```python
print("Transpose of A:\n", A.T) #Swaps rows and columns
```

A system of linear equations can be written in matrix form as:

Ax=b

Where:

● A is the coefficient matrix,

● x is the vector of unknowns,

● b is the constant vector.


```python
#2x + 3y = 5
#4x - y = 3
# [[2, 3],     [[x]       [[5]
#  [4, -1]]     [y]]   =   [3]]

# Coefficient matrix A and constant vector b
A = np.array([[2, 3], [4, -1]])
b = np.array([5, 3])
# Solve the system A * x = b
x = np.linalg.solve(A, b)
print("Solution (x, y):", x)
```

Singular Value Decomposition (SVD) is a factorization of a matrix into three
matrices:

A = UΣV^T

U & V are orthogonal matrices

Σ is a diagonal matrix of singular values

```python
A = np.array([[1, 2], [3, 4], [5, 6]])

U, sigma, Vt = np.linalg.svd(A)

print("U matrix:\n", U)
print("Sigma values:", sigma)
print("V^T matrix:\n", Vt)
```
