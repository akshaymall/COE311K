```python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D
```

```python
x = np.linspace(0, 2 * np.pi, 50)

plt.subplot(2, 2, 1) # (row, column, active area)
plt.plot(x, np.sin(x), 'r') # will go into active area #1 

plt.subplot(2, 2, 2)
plt.plot(x, np.cos(x), 'g') # will go into active area #2

plt.subplot(2, 2, 3)
def f1(x,y):
    return (x*x+y*y)
x_mesh = np.linspace(-10, 10, 2000)
y_mesh = np.linspace(-10, 10, 2000)
X, Y = np.meshgrid(x_mesh, y_mesh)
Z = f1(X, Y)


plt.subplot(2, 2, 4)
x_scatter = np.random.rand(1000)
y_scatter = np.random.rand(1000)
size = np.random.rand(1000) * 50
color = np.random.rand(1000)
plt.scatter(x_scatter, y_scatter, size, color)
plt.colorbar()
plt.show()
```
